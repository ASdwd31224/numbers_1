```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(digit.data, digit.target, test_size=0.3)
```

```python
X_train[0]
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,1))
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train[0]
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

def all_metrict(y_train, y_test, y_predict_train, y_predict_test) :
    acc_train = accuracy_score(y_true=y_train, y_pred=y_predict_train)
    acc_test = accuracy_score(y_true=y_test, y_pred=y_predict_test)
    
    pre_test = precision_score(y_true=y_test, y_pred=y_predict_test, average="weighted")
    rec_test = recall_score(y_true=y_test, y_pred=y_predict_test, average="weighted")
    
    print(f"Acc train: {acc_train} ===> Acc test: {acc_test} ===> Precision: {pre_test} ===> Recall: {rec_test}")
    
    return acc_train, acc_test, pre_test, rec_test
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(max_depth=128, n_estimators=256)
rf.fit(X_train, y_train)
```

```python
y_predict_train = rf.predict(X_train)
y_predict_test = rf.predict(X_test)

acc_train_rf, acc_test_rf, pre_test_rf, rec_test_rf = all_metrict(y_train, y_test, y_predict_train, y_predict_test)
```

```python
from sklearn.svm import SVC

svm = SVC(kernel="poly")
svm.fit(X_train, y_train)
```

```python
y_predict_train = svm.predict(X_train)
y_predict_test = svm.predict(X_test)

acc_train_svm, acc_test_svm, pre_test_svm, rec_test_svm = all_metrict(y_train, y_test, y_predict_train, y_predict_test)
```

```python
from sklearn.neural_network import MLPClassifier

ann = MLPClassifier(hidden_layer_sizes=256, solver="adam", learning_rate="adaptive", batch_size=64)
ann.fit(X_train, y_train)
```

```python
y_predict_train = ann.predict(X_train)
y_predict_test = ann.predict(X_test)

acc_train_ann, acc_test_ann, pre_test_ann, rec_test_ann = all_metrict(y_train, y_test, y_predict_train, y_predict_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(X_train, y_train)
```

```python
y_predict_train = knn.predict(X_train)
y_predict_test = knn.predict(X_test)

acc_train_knn, acc_test_knn, pre_test_knn, rec_test_knn = all_metrict(y_train, y_test, y_predict_train, y_predict_test)
```

```python
import matplotlib.pyplot as plt

X_test_plt = [acc_test_rf, acc_test_svm, acc_test_knn, acc_test_ann]
title = ["RF", "SVM", "KNN", "ANN"]
# colors = ["red", "blue", "green", "yellow"]
plt.bar(title, X_test_plt, color="black")  # can use colors
plt.grid()
plt.show()
```
